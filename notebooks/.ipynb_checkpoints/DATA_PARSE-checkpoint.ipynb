## Парсинг публикаций

### Написание функции по извлечению данных публикаций с помощью VK API

```python
pip install vk_api
```

```python
api_token = #'Вставьте свой ключ доступа'
# используемый личный ключ скрыт из ячейки в целях безопасности аккаунта в соцсети
```

```python
# импорт библиотек
import vk_api
from datetime import datetime
import time
import pandas as pd

# инициализация функции для извлечения публикаций
def get_posts(api_token, group_id, count=100, start_offset=5000):
  '''
  Получает тексты постов и пользовательские реакции из сообществ с сайта https://vk.com.
  Собирает посты, пока не дойдет до даты 01.01.2020, либо пока не закончатся посты.

  Параметры:
    api_token: ключ доступа для выполнения запросов к VK API
    group_id: уникальный идентификатор сообщества, начинающийся с '-'
    count: число выгружаемых постов за 1 итерацию
    start_offset: начальная позиция со смещением (сколько последних постов пропускаем перед сбором данных)
  Вывод функции:
    Возвращает all_posts – список из словарей; в каждом словаре содержится информация
                           о конкретной публикации.
  '''
  # инициализация VK API
  vk_session = vk_api.VkApi(token=api_token)
  vk = vk_session.get_api()

  # инициализация словаря, смещения, целевой даты и флага ее достижения
  all_posts = []
  offset = start_offset
  target_date = datetime(2020, 1, 1).timestamp()
  target_flag = False

  print(f'Начинаем сбор с offset = {offset}')
  print(f'Цель: дойти до 01.01.2020')
  print('-' * 50)

  # пока не достигнута целевая дата, цикл в работе
  while not target_flag:
    # отправка запроса к новостной 'стене' сообщества
    response = vk.wall.get(owner_id=group_id,
                           filter='owner',
                           count=count,
                           offset=offset)
    # пустой ответ на запрос сигнализирует об отсутствии постов
    if not response.get('items'):
      print('Постов больше нет')
      break
    # извлечение даты поста из ответа на запрос
    for post in response['items']:
      post_date = post['date']
      post_date_string = datetime.fromtimestamp(post_date).strftime('%Y-%m-%d')
      post_year = datetime.fromtimestamp(post_date).year
      # проверка на достижении целевой даты 01.01.2020
      if post_date <= target_date:
        target_flag = True
        if post_year == 2020 or post_year < 2020:
          print(f'\n- Достигли {post_year} года -')
          print(f'Последний пост: {post['id']} от {post_date_string}')
          break
      # извлечение данных из ответа на запрос в виде словаря и присоединение их к списку со всеми постами
      else:
        all_posts.append({'post_id': post['id'],
                          'date': post_date_string,
                          'text': post['text'],
                          'likes': post.get('likes', {}).get('count', 0),
                          'comments_count': post.get('comments', {}).get('count', 0),
                          'reposts': post.get('reposts', {}).get('count', 0),
                          'views': post.get('views', {}).get('count', 0) if 'views' in post else 0})
    # завершение цикла в случае достижения целевой даты 01.01.2020
    if target_flag:
      break

    print(f'Обработано offset {offset}, собрано {len(all_posts)} постов (последний: {post_date_string})')
    # если цикл продолжается, значение смещения обновляется на число собранных постов (100)
    offset += 100
    # задержка алгоритма для избежания превышения установленных лимитов запросов (rps)
    time.sleep(0.1)

  return all_posts
```

Сообщества университетов Москвы, имеющие суммарный индекс > 1. Сайт: https://brandanalytics.ru/university-rankings/integral/#scroll_anchor

### Агрегация данных в единый датафрейм

```python
import os

# все файлы были перенесены в отдельную папку, откуда осуществляется их выгрузка
folder_path = '/Users/ekaterinamalkova/Desktop/vkr_analysis/data/data_parsed_unis_csvis'
dataframes = []

# последовательное чтение каждого файла и их присоединение к списку из Dataframe'ов
for file in os.listdir(folder_path):
  file_path = os.path.join(folder_path, file)
  df = pd.read_csv(file_path)
  df['source_nickname'] = file
  print(f'Dataframe {file} содержит {len(df)} публикаций, присоединяем их к основному Dataframe')
  dataframes.append(df)

# преобразование списка из Dataframe'ов в основной Datarame
if dataframes:
  data = pd.concat(dataframes, ignore_index=True)
  data['date'] = pd.to_datetime(data['date'])
  print('')
  print(f'Все готово, основной Dataframe содержит {len(data)} публикаций с 2020-01-01')
  print('')
  data = data.query("date < '2026-01-01'").reset_index(drop=True)
  data = data.dropna()
  print(f'Удалены публикации из 2026 и пропущенные значения, сновной Dataframe содержит {len(data)} публикаций с 2020-01-01 по 2025-12-31')

data.to_csv('/Users/ekaterinamalkova/Desktop/vkr_analysis/data/data_unis.csv', index=False)
```

### Написание функции по извлечению данных комментариев с помощью VK API

```python
# импорт библиотек
import vk_api
from datetime import datetime
import time
import pandas as pd

# инициализация функции для извлечения комментариев
def get_comments_from_posts(api_token, post_link):
    '''
    Получает тексты комментариев из публикаций сообществ с сайта https://vk.com.

    Параметры:
        api_token: ключ доступа для выполнения запросов к VK API
        post_link: полная ссылка на публикацию
    Вывод функции:
        Возвращает str – тексты всех комментариев к публикации, через разделитель '|'; '0' – если комментарии отсутсвуют
    '''
  # инициализация VK API
    vk_session = vk_api.VkApi(token=api_token)
    vk = vk_session.get_api()

    wall_ = post_link.split('wall')[1]
    owner_id, post_id = wall_.split('_')
    owner_id = int(owner_id)
    post_id = int(post_id)

    # инициализируем вспомогательные элементы
    all_comments = []
    offset = 0
    count = 100

    # пока все комментарии не собраны, выполняется цикл
    while True:
        # отправка запроса для получения кромментариев
        comments_response = vk.wall.getComments(owner_id=owner_id,
                                               post_id=post_id,
                                               count=count,
                                               offset=offset,
                                               need_likes=0,
                                               extended=0)
        
        if comments_response.get('items'):
            for comment in comments_response['items']:
                comment_text = comment.get('text', '')
                if comment_text and comment_text.strip():
                    all_comments.append(comment_text.strip())


            if len(comments_response['items']) < count:
                break
                
        offset += count
        # задержка алгоритма для избежания превышения установленных лимитов запросов (rps)
        time.sleep(0.05)
    # объединение комментариев в ячейку
    if all_comments:
        text = ' | '.join(all_comments)
        return text
    else:
        return '0'
```

```python

import re
import time
import threading
import warnings

import vk_api
import pandas as pd
from tqdm import tqdm


warnings.filterwarnings("ignore")


# =========================
# Настройки
# =========================

PROGRESS_FILE = "progress_3.csv"
FINAL_FILE = "final_result_3.csv"

TIMEOUT_SECONDS = 80
SAVE_EVERY = 100

CHECK_TEXT = "нужна проверка"


# =========================
# Функции
# =========================

def parse_wall_link(post_link):
    match = re.search(r"wall(-?\d+)_(\d+)", str(post_link))

    if not match:
        raise ValueError("bad post link")

    owner_id = int(match.group(1))
    post_id = int(match.group(2))

    return owner_id, post_id


def get_comments_from_posts(api_token, post_link):
    try:
        vk_session = vk_api.VkApi(token=api_token)
        vk = vk_session.get_api()

        owner_id, post_id = parse_wall_link(post_link)

        all_comments = []
        offset = 0
        count = 100

        while True:
            comments_response = vk.wall.getComments(
                owner_id=owner_id,
                post_id=post_id,
                count=count,
                offset=offset,
                need_likes=0,
                extended=0
            )

            items = comments_response.get("items", [])

            for comment in items:
                comment_text = comment.get("text", "")
                if comment_text and comment_text.strip():
                    all_comments.append(comment_text.strip())

            if len(items) < count:
                break

            offset += count
            time.sleep(0.08)

        if all_comments:
            return " | ".join(all_comments)

        return "0"

    except Exception:
        return CHECK_TEXT


def get_comments_safe(api_token, post_link, timeout=80):
    result = {"value": CHECK_TEXT}

    def target():
        try:
            result["value"] = get_comments_from_posts(api_token, post_link)
        except Exception:
            result["value"] = CHECK_TEXT

    thread = threading.Thread(target=target)
    thread.daemon = True
    thread.start()
    thread.join(timeout)

    if thread.is_alive():
        return CHECK_TEXT

    if result["value"]:
        return result["value"]

    return CHECK_TEXT


# =========================
# Подготовка данных
# =========================

data_with_comments_3 = (
    data_with_comments
    .iloc[50000:]
    .reset_index(drop=True)
)


try:
    progress_df = pd.read_csv(PROGRESS_FILE)
    already_done = len(progress_df)
    comments = progress_df["comments"].fillna(CHECK_TEXT).tolist()
except FileNotFoundError:
    already_done = 0
    comments = []


# =========================
# Основной цикл
# =========================

for i, post_link in enumerate(
    tqdm(
        data_with_comments_3["post_link"].iloc[already_done:],
        desc="Сбор комментариев",
        initial=already_done,
        total=len(data_with_comments_3)
    )
):
    current_index = already_done + i

    comment = get_comments_safe(
        api_token=api_token,
        post_link=post_link,
        timeout=TIMEOUT_SECONDS
    )

    comments.append(comment)

    time.sleep(0.15)

    if (current_index + 1) % SAVE_EVERY == 0:
        temp_df = data_with_comments_3.iloc[:current_index + 1].copy()
        temp_df["comments"] = comments
        temp_df.to_csv(PROGRESS_FILE, index=False)


# =========================
# Финальное сохранение
# =========================

result_df = data_with_comments_3.iloc[:len(comments)].copy()
result_df["comments"] = comments

result_df.to_csv(PROGRESS_FILE, index=False)
result_df.to_csv(FINAL_FILE, index=False)

print("Готово!")
print(f"Итоговый файл: {FINAL_FILE}")
```

```python
import re
import time
import threading
import warnings

import vk_api
import pandas as pd
from tqdm import tqdm


warnings.filterwarnings("ignore")


# =========================
# Настройки
# =========================

CHECK_TEXT = "нужна проверка"

PROGRESS_FILE = "fixed_errors_progress.csv"
FINAL_FILE = "data_with_comments_fixed_errors.csv"

TIMEOUT_SECONDS = 120
SAVE_EVERY = 20

SHOW_EACH_POST = True


# =========================
# Исправляем ссылки только на срезе с ошибками
# =========================

error_mask = data_with_comments["comments"].eq(CHECK_TEXT)

bad_club_mask = (
    error_mask
    & data_with_comments["post_link"].astype(str).str.contains("vk.com/club-", na=False)
)

data_with_comments.loc[bad_club_mask, "post_link"] = (
    data_with_comments.loc[bad_club_mask, "post_link"]
    .astype(str)
    .str.replace("vk.com/club-", "vk.com/club", regex=False)
)

print(f"Исправлено ссылок с лишним минусом после club: {bad_club_mask.sum()}")


# =========================
# Функции
# =========================

def parse_wall_link(post_link):
    match = re.search(r"wall(-?\d+)_(\d+)", str(post_link))

    if not match:
        raise ValueError("bad post link")

    owner_id = int(match.group(1))
    post_id = int(match.group(2))

    return owner_id, post_id


def get_comments_from_posts(api_token, post_link, verbose=False):
    try:
        vk_session = vk_api.VkApi(token=api_token)
        vk = vk_session.get_api()

        owner_id, post_id = parse_wall_link(post_link)

        all_comments = []
        offset = 0
        count = 100
        page_num = 0

        while True:
            page_num += 1

            comments_response = vk.wall.getComments(
                owner_id=owner_id,
                post_id=post_id,
                count=count,
                offset=offset,
                need_likes=0,
                extended=0
            )

            items = comments_response.get("items", [])
            total_count = comments_response.get("count", 0)

            for comment in items:
                comment_text = comment.get("text", "")
                if comment_text and comment_text.strip():
                    all_comments.append(comment_text.strip())

            if verbose:
                tqdm.write(
                    f"   страница {page_num}: получено {len(items)}, "
                    f"накоплено текстовых {len(all_comments)}, всего по API {total_count}"
                )

            if len(items) < count:
                break

            offset += count
            time.sleep(0.08)

        return " | ".join(all_comments) if all_comments else "0"

    except Exception as e:
        if verbose:
            tqdm.write(f"   ошибка: {type(e).__name__}: {e}")
        return CHECK_TEXT


def get_comments_safe(api_token, post_link, timeout=120, verbose=False):
    result = {"value": CHECK_TEXT}

    def target():
        try:
            result["value"] = get_comments_from_posts(
                api_token=api_token,
                post_link=post_link,
                verbose=verbose
            )
        except Exception as e:
            if verbose:
                tqdm.write(f"   ошибка потока: {type(e).__name__}: {e}")
            result["value"] = CHECK_TEXT

    thread = threading.Thread(target=target)
    thread.daemon = True
    thread.start()
    thread.join(timeout)

    if thread.is_alive():
        if verbose:
            tqdm.write(f"   таймаут {timeout} сек")
        return CHECK_TEXT

    return result["value"] if result["value"] else CHECK_TEXT


# =========================
# Повторный парсинг только строк с ошибками
# =========================

error_indices = data_with_comments.index[
    data_with_comments["comments"].eq(CHECK_TEXT)
].tolist()

print(f"Строк для повторного парсинга: {len(error_indices)}")


for n, idx in enumerate(
    tqdm(error_indices, desc="Повторный сбор комментариев"),
    start=1
):
    row = data_with_comments.loc[idx]

    post_link = row["post_link"]
    source = row.get("source_nickname", "")
    post_id = row.get("post_id", "")
    comments_count = row.get("comments_count", "")

    if SHOW_EACH_POST:
        tqdm.write("")
        tqdm.write(
            f"[{n}/{len(error_indices)}] "
            f"idx={idx}, source={source}, post_id={post_id}, "
            f"comments_count={comments_count}"
        )
        tqdm.write(f"   link: {post_link}")

    new_comment = get_comments_safe(
        api_token=api_token,
        post_link=post_link,
        timeout=TIMEOUT_SECONDS,
        verbose=SHOW_EACH_POST
    )

    data_with_comments.at[idx, "comments"] = new_comment

    if SHOW_EACH_POST:
        if new_comment == CHECK_TEXT:
            tqdm.write("   результат: нужна проверка")
        elif new_comment == "0":
            tqdm.write("   результат: комментариев нет / текстовых комментариев нет")
        else:
            tqdm.write(f"   результат: успешно, длина текста = {len(new_comment)} символов")

    time.sleep(0.15)

    if n % SAVE_EVERY == 0:
        data_with_comments.to_csv(PROGRESS_FILE, index=False)
        tqdm.write(f"   сохранено после {n} повторно обработанных строк")


# =========================
# Сохранение
# =========================

data_with_comments.to_csv(PROGRESS_FILE, index=False)
data_with_comments.to_csv(FINAL_FILE, index=False)

print("Готово!")
print(f"Осталось строк с ошибкой: {(data_with_comments['comments'] == CHECK_TEXT).sum()}")
print(f"Итоговый файл: {FINAL_FILE}")
```

```python
import re
import time
import threading
import warnings

import vk_api
import pandas as pd
from tqdm import tqdm


warnings.filterwarnings("ignore")


# =========================
# Загружаем актуальный файл
# =========================

data_with_comments = pd.read_csv("data_with_comments_fixed_errors.csv")


# =========================
# Настройки
# =========================

CHECK_TEXT = "нужна проверка"

PROGRESS_FILE = "data_with_comments_fixed_errors_progress.csv"
FINAL_FILE = "data_with_comments_fixed_errors.csv"

TIMEOUT_SECONDS = 120
SAVE_EVERY = 20


# =========================
# Исправляем известный group_id
# =========================

FIX_GROUP_IDS = {
    "surikov.csv": "-194894715"
}


for source_name, group_id in FIX_GROUP_IDS.items():
    mask = data_with_comments["source_nickname"].eq(source_name)

    group_id_str = str(group_id)
    club_id_str = group_id_str.replace("-", "")

    data_with_comments.loc[mask, "group_id"] = group_id_str

    data_with_comments.loc[mask, "post_link"] = (
        "https://vk.com/club"
        + club_id_str
        + "?w=wall"
        + group_id_str
        + "_"
        + data_with_comments.loc[mask, "post_id"].astype(str)
    )

print("После исправления group_id:")
print(data_with_comments[data_with_comments["source_nickname"].eq("surikov.csv")][
    ["source_nickname", "group_id", "post_id", "post_link", "comments"]
])


# =========================
# Функции сбора комментариев
# =========================

def parse_wall_link(post_link):
    match = re.search(r"wall(-?\d+)_(\d+)", str(post_link))

    if not match:
        raise ValueError("bad post link")

    owner_id = int(match.group(1))
    post_id = int(match.group(2))

    return owner_id, post_id


def get_comments_from_posts(api_token, post_link):
    try:
        vk_session = vk_api.VkApi(token=api_token)
        vk = vk_session.get_api()

        owner_id, post_id = parse_wall_link(post_link)

        all_comments = []
        offset = 0
        count = 100
        page = 1

        tqdm.write(f"Старт поста: {post_link}")

        while True:
            comments_response = vk.wall.getComments(
                owner_id=owner_id,
                post_id=post_id,
                count=count,
                offset=offset,
                need_likes=0,
                extended=0
            )

            items = comments_response.get("items", [])

            for comment in items:
                comment_text = comment.get("text", "")
                if comment_text and comment_text.strip():
                    all_comments.append(comment_text.strip())

            tqdm.write(
                f"Пост {post_id}: страница {page}, "
                f"получено {len(items)}, всего текстов {len(all_comments)}"
            )

            if len(items) < count:
                break

            offset += count
            page += 1
            time.sleep(0.08)

        tqdm.write(f"Готов пост {post_id}: итог {len(all_comments)} комментариев")

        return " | ".join(all_comments) if all_comments else "0"

    except Exception as e:
        tqdm.write(f"Ошибка по посту {post_link}: {e}")
        return CHECK_TEXT


def get_comments_safe(api_token, post_link, timeout=120):
    result = {"value": CHECK_TEXT}

    def target():
        try:
            result["value"] = get_comments_from_posts(api_token, post_link)
        except Exception:
            result["value"] = CHECK_TEXT

    thread = threading.Thread(target=target)
    thread.daemon = True
    thread.start()
    thread.join(timeout)

    if thread.is_alive():
        tqdm.write(f"Таймаут по посту: {post_link}")
        return CHECK_TEXT

    return result["value"] if result["value"] else CHECK_TEXT


# =========================
# Повторная обработка ошибок
# =========================

error_mask = data_with_comments["comments"].eq(CHECK_TEXT)
error_indices = data_with_comments.index[error_mask].tolist()

print(f"Найдено строк для повторной обработки: {len(error_indices)}")


for n, idx in enumerate(
    tqdm(error_indices, desc="Повторная обработка ошибок"),
    start=1
):
    post_link = data_with_comments.at[idx, "post_link"]

    tqdm.write(f"\n[{n}/{len(error_indices)}] Обработка строки index={idx}")

    new_comment = get_comments_safe(
        api_token=api_token,
        post_link=post_link,
        timeout=TIMEOUT_SECONDS
    )

    data_with_comments.at[idx, "comments"] = new_comment

    time.sleep(0.15)

    if n % SAVE_EVERY == 0:
        data_with_comments.to_csv(PROGRESS_FILE, index=False)
        tqdm.write(f"Промежуточно сохранено: {n} обработанных ошибок")


# =========================
# Финальное сохранение
# =========================

data_with_comments.to_csv(PROGRESS_FILE, index=False)
data_with_comments.to_csv(FINAL_FILE, index=False)

print("Готово!")
print(f"Осталось строк с ошибкой: {(data_with_comments['comments'] == CHECK_TEXT).sum()}")
print(f"Итоговый файл: {FINAL_FILE}")
```
